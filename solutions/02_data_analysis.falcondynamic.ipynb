# 5.Noise Filtering

```python
# Define criteria for noise filtering
def filter_noise(tokens, min_freq=2): 
    # Count the frequency of each token
    token_counts = Counter(tokens)
    # Remove tokens with fewer than three characters and those below the minimum frequency
    filtered = [token for token in tokens if len(token) >= 3 and token_counts[token] >= min_freq]
    return filtered
```

```python
# Apply noise filtering
filtered_stemmed_tokens = filter_noise(stemmed_tokens)
filtered_lemmatized_tokens = filter_noise(lemmatized_tokens)
```

```python
# Print the filtered tokens
print("Filtered Stemmed Tokens:", filtered_stemmed_tokens)
print("Filtered Lemmatized Tokens:", filtered_lemmatized_tokens)
```

# Final Validation and Testing

```python
# Define a function for the preprocessing pipeline
def preprocessing_pipeline(content):
    
    # convert tect to lowercase
    content = content.lower().strip()

    # Divide lines
    lines = content.split('\n')   
  
    #removing Human 1: or Human 2: from the begining of the line
    lines = [re.sub(r"^human \d+: ", "", line) for line in lines if line.strip() != ""]
    
    content_clean = ' '.join(lines)
    
    # Remove unnecessary characters and numbers
    content_clean = re.sub(r'[^a-z\s]', '', content_clean) 
    
    # Strip out URLs, email addresses and usernames
    content_clean = re.sub(r'http\S+|www\S+|@\S+', '', content_clean)
    
    #Removes extra spaces caused by removing unnecessary characters
    content_clean = re.sub(r'\s+', ' ', content_clean)
    
    # Tokenize the cleaned text
    tokens = word_tokenize(content_clean)
    
    # Load the predefined stop words list for English
    stop_words = set(stopwords.words('english')) 
    
    # Customize the stop words list if needed
    custom_stop_words = set(["additional", "words"])
    stop_words.update(custom_stop_words)
    
    # Remove stop words from the token list
    filtered_tokens = [token for token in tokens if token not in stop_words]
   
   # Initialize stemmer and lemmatizer
    stemmer = SnowballStemmer('english')

    # Perform stemming
    stemmed_tokens = [stemmer.stem(token) for token in filtered_tokens]

    # Perform lemmatization
    lemmatized_tokens = [lemmatizer.lemmatize(token) for token in filtered_tokens]
    
    # Apply noise filtering
    filtered_stemmed_tokens = filter_noise(stemmed_tokens)
    filtered_lemmatized_tokens = filter_noise(lemmatized_tokens)
    
    return filtered_stemmed_tokens, filtered_lemmatized_tokens
        
```

```python
# Sample data for testing
sample_data = [
    "Hi! This is a test.",
    "Empty text.",
    "Highly noisy input!!! $$$%%%***###",
    "Another example with numbers: 123 456",
    "Mix of English and Türkçe metin: Hello, nasılsın?"
]
```

```python
# Test the preprocessing pipeline with sample data
for i, text in enumerate(sample_data):
    stemmed, lemmatized = preprocessing_pipeline(text)
    print(f"Sample {i+1}:")
    print(f"Original Text: {text}")
    print(f"Stemmed Tokens: {stemmed}")
    print(f"Lemmatized Tokens: {lemmatized}")
    print("-" * 50)
```

```python
assert preprocessing_pipeline("") == ([], []), "Edge case for empty text failed"
assert preprocessing_pipeline("Highly noisy input!!! $$$%%%***###") == ([], []), "Edge case for highly noisy input failed"
```
